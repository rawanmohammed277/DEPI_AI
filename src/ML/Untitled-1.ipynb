B6 WRITE CODE A balance starts at 1000 and grows by 10 % every year. Using a while loop, print the number of
whole years needed until the balance is at least 2000. (Expected output: a single integer.)

```python
balance= 1000
years=0
while balance < 2000:
    balance*= 1.10
    years+=1
print(years)
```

```python
C7 WRITE CODE Given nums = [3, 8, 12, 5, 6, 7], use a loop and append() to build a new list containing only
the even numbers. Printing it must show [8, 12, 6].
```

```python
nums = [3, 8, 12, 5, 6, 7]
even_nums= []
for num in nums:
    if num%2==0:
        even_nums.append(num)
print(even_nums)
```
