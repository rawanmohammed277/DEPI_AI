```python
score = float(input("enter your degree"))
if 0 <= score <= 100:
    if score >= 90:
        grade = 'A'
    elif score >= 80:
        grade = 'b'
    elif score >= 70:
        grade = 'c'
    elif score >= 60:
        grade = 'd'
    else:
        grade = 'f'

    print(f"the grade is : {grade}")
else:
    print(f"invalid degree")
```
