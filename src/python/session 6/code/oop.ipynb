```python
from enum import member


class Member:
    
    not_allowed_name= ['hell', 'shit' , 'baloot']
    user_num = 0
    
    @classmethod
    def show_users_count(cls):
        print(f'we have {cls.user_num} users in our system ')
        
    
    
    def __init__(self, first_name, middle_name, last_name, gender):
        self.fname = first_name
        self.mname = middle_name
        self.lname = last_name
        self.gender = gender
        Member.user_num +=1
        
    def full_name(self):
        if self.fname in Member.not_allowed_name:
            raise ValueError("Name Not Allowed")
        else:
            return f'{self.fname} {self.mname} {self.lname}'
    
    def name_with_title(self):
        
        if self.gender == "Female":
            return f'Hello Mrs {self.fname}'
        
        elif self.gender == "Male":
            return f'Hello Mr {self.fname}'
        else :
            return f'Hello {self.fname}'
            
        
    def get_all_info(self):
        return f'{self.name_with_title()}, your full name is {self.full_name()}'
    
    
    def delete_user(self):
        Member.user_num -=1
        return f'User {self.fname} Deleted'
```

```python
print(Member.user_num)
m1=Member("Rawan","A","Mohamed","Female")
m2=Member("Shahd","A","Mohamed","Male")
m3=Member("Aliaa","A","Mohamed","Male")
m4=Member("shit","A","Mohamed","Male")

print(Member.user_num)

#print(dir(m1))

print(m1.fname,m1.mname,m1.lname)

print(m2.fname)
print('---------------')
print(m1.full_name())
print(m1.name_with_title())
print(m2.name_with_title())
print('---------------')
print(m1.get_all_info())
print(m2.get_all_info())

print('---------------')
print(m4.delete_user())
print(Member.user_num)

Member.show_users_count()

print(m1.full_name())
```

```python
class Food:# base class
    def __init__(self,name):
        self.name= name
        print(f"{self.name} is created from main class")
        
    def eat(self):
        print("eat method from base class")
        
class Apple(Food):
    def __init__(self):
        print("Apple is created from derived class")
        
    
```

```python
f1=Food("pizza")
f2=Apple()
f2.eat()
```
