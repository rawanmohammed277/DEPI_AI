## **Func1:  multiplication_table**
    


```python
class MultiplicationTable:
    """
    MultiplicationTable class handles the operation
    of printing the multiplication table of a number.

    Attributes:
        number (int): The number whose multiplication table will be printed.
    """

    def __init__(self, number):
        self.number = number

    def multiplication_table(self):
        """
        Prints the multiplication table of the stored number
        from 1 to 10.

        Returns:
            None
        """
        for i in range(1, 11):
            print(f"{self.number} x {i} = {self.number * i}")


# Get input from the user
number = int(input("Enter a number: "))

# Create an object
table = MultiplicationTable(number)

# Call the main method
table.multiplication_table()
```

## **Func2: Twin Primes**

```python
class TwinPrimes:
    """
    TwinPrimes class handles finding and printing
    twin prime numbers within a given limit.

    Attributes:
        limit (int): The upper limit for finding twin primes.
    """

    def __init__(self, limit=1000):
        self.limit = limit

    def is_prime(self, n):
        """
        Checks whether a number is prime.

        Parameters:
            n (int): The number to be checked.

        Returns:
            bool: True if the number is prime, otherwise False.
        """
        if n < 2:
            return False

        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                return False

        return True

    def print_twin_primes(self):
        """
        Finds and prints all twin prime pairs
        less than the given limit.

        Returns:
            None
        """
        print(f"Twin primes less than {self.limit}:")

        # Check odd numbers for twin prime pairs
        for p in range(3, self.limit - 2, 2):
            if self.is_prime(p) and self.is_prime(p + 2):
                print(f"({p}, {p + 2})")


# Create an object
twin = TwinPrimes(1000)

# Call the main method
twin.print_twin_primes()
```

## **Func3: Prime Factors**

```python
class PrimeFactors:
    """
    PrimeFactors class handles finding the prime factors
    of a given number.

    Attributes:
        number (int): The number whose prime factors will be found.
    """

    def __init__(self, number):
        self.number = number

    def get_prime_factors(self):
        """
        Finds the prime factors of the stored number.

        Returns:
            list: A list containing the prime factors of the number.
        """
        factors = []
        n = self.number

        # Extract all factors of 2
        while n % 2 == 0:
            factors.append(2)
            n //= 2

        divisor = 3

        # Check odd divisors
        while divisor * divisor <= n:
            while n % divisor == 0:
                factors.append(divisor)
                n //= divisor

            divisor += 2

        # If n is greater than 1, it is a prime factor
        if n > 1:
            factors.append(n)

        return factors


try:
    number = int(input("Enter a number to find its prime factors: "))

    if number <= 1:
        print("Please enter a number greater than 1.")
    else:
        # Create an object
        prime_factors = PrimeFactors(number)

        # Call the method
        result = prime_factors.get_prime_factors()

        print(
            f"The prime factors of {number} are: "
            f"{', '.join(map(str, result))}"
        )

except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

## **Func4: converts a decimal number to binary numbe**

```python
class Binary:
    """
    Binary class handles converting a decimal number
    into its binary representation.

    Attributes:
        number (int): The decimal number to be converted.
    """

    def __init__(self, number):
        self.number = number

    def binary(self):
        """
        Converts the stored decimal number into binary.

        Returns:
            str: The binary representation of the number.
        """
        if self.number == 0:
            return "0"

        number = self.number
        reminders = []

        # Store the remainders of division by 2
        while number > 0:
            rem = number % 2
            reminders.append(str(rem))
            number //= 2

        # Reverse the remainders to get the correct binary number
        return ''.join(reminders[::-1])


# Get input from the user
num = int(input("Enter a number: "))

# Create an object
binary_number = Binary(num)

# Call the method and print the result
result = binary_number.binary()
print(f"Binary representation of {num}: {result}")
```

## **Func5: Perfect Numbers**

```python
class PerfectNumbers:
    """
    PerfectNumbers class handles finding perfect numbers
    within a given range.

    Attributes:
        start (int): The starting value of the range.
        end (int): The ending value of the range.
    """

    def __init__(self, start, end):
        self.start = start
        self.end = end

    def is_perfect(self, n):
        """
        Checks whether a number is a perfect number.

        Parameters:
            n (int): The number to be checked.

        Returns:
            bool: True if the number is perfect, otherwise False.
        """
        if n < 2:
            return False

        divisors_sum = 1

        # Find and add all proper divisors
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                divisors_sum += i

                # Avoid adding the same divisor twice for perfect squares
                if i != n // i:
                    divisors_sum += n // i

        return divisors_sum == n

    def find_perfect_numbers(self):
        """
        Finds all perfect numbers within the specified range.

        Returns:
            list: A list containing the perfect numbers in the range.
        """
        perfect_numbers = []

        # Check every number in the given range
        for num in range(self.start, self.end + 1):
            if self.is_perfect(num):
                perfect_numbers.append(num)

        return perfect_numbers


# Get the range from the user
start_range = int(input("Enter start of range: "))
end_range = int(input("Enter end of range: "))

# Create an object
perfect = PerfectNumbers(start_range, end_range)

# Call the main method
result = perfect.find_perfect_numbers()

# Print the result
print(f"Perfect numbers between {start_range} and {end_range}: {result}")
```

## **Func6: Calculator**

```python
class Calculator:
    """
    Calculator class performs basic mathematical operations
    including addition, subtraction, multiplication, and division.
    """

    def __init__(self):
        """
        Initializes the Calculator object.
        """
        pass

    def get_number(self, prompt_message):
        """
        Prompts the user for numerical input and handles
        invalid non-numeric entries.

        Parameters:
            prompt_message (str): The message displayed to the user.

        Returns:
            float: A valid numerical value.
        """
        while True:
            try:
                return float(input(prompt_message))
            except ValueError:
                print("Invalid input! Please enter a valid number.")

    def calculate(self, op_choice, num1, num2):
        """
        Performs the selected mathematical operation on two numbers.

        Parameters:
            op_choice (str): The selected operation (1-4).
            num1 (float): The first number.
            num2 (float): The second number.

        Returns:
            tuple: The calculated result and operation name.

        Raises:
            ZeroDivisionError: If division by zero is attempted.
        """
        if op_choice == '1':
            return num1 + num2, "adding"

        elif op_choice == '2':
            return num1 - num2, "subtracting"

        elif op_choice == '3':
            return num1 * num2, "multiplying"

        elif op_choice == '4':
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero.")

            return num1 / num2, "dividing"

    def format_num(self, val):
        """
        Formats floating-point numbers as integers when
        they have no fractional part.

        Parameters:
            val (float): The number to format.

        Returns:
            int or float: The formatted number.
        """
        return int(val) if val.is_integer() else val

    def main(self):
        """
        Runs the main calculator loop and manages
        user interaction.
        """
        while True:
            print("\n=== Basic Calculator ===")
            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")
            print("5. Exit")

            choice = input("Select an option (1-5): ").strip()

            # Exit the calculator
            if choice == '5':
                print("Thank you for using the calculator! Goodbye.")
                break

            # Validate the menu choice
            if choice not in ['1', '2', '3', '4']:
                print("Invalid selection! Please choose an option between 1 and 5.")
                continue

            n1 = self.get_number("Enter the first number: ")
            n2 = self.get_number("Enter the second number: ")

            try:
                result, operation_name = self.calculate(choice, n1, n2)

                print(
                    f"The result of {operation_name} "
                    f"{self.format_num(n1)} and {self.format_num(n2)} "
                    f"is {self.format_num(result)}."
                )

            except ZeroDivisionError:
                print("Error: Division by zero is not allowed.")

            again = input(
                "\nDo you want to perform another calculation? (yes/no): "
            ).strip().lower()

            if again not in ['yes', 'y']:
                print("Thank you for using the calculator! Goodbye.")
                break


# Create an object
calculator = Calculator()

# Call the main method
calculator.main()
```

## **Func7: ThanosSnap**

```python
import os
import random


class ThanosSnap:
    """
    ThanosSnap class simulates the 'Thanos Snap' operation
    by creating files in a folder and randomly deleting half of them.

    Attributes:
        folder_path (str): The path of the target folder.
        num_files_to_create (int): The number of files to create.
    """

    def __init__(self, folder_path, num_files_to_create=50):
        """
        Initializes the ThanosSnap object.

        Parameters:
            folder_path (str): The path of the target folder.
            num_files_to_create (int): The number of files to create.
        """
        self.folder_path = folder_path
        self.num_files_to_create = num_files_to_create

    def thanos_snap(self):
        """
        Creates files in the target folder and randomly deletes
        half of the files.

        Returns:
            tuple: A tuple containing the initial and final file counts.
        """

        # Ensure the target folder exists
        os.makedirs(self.folder_path, exist_ok=True)

        # Create the specified number of files
        for i in range(1, self.num_files_to_create + 1):
            file_path = os.path.join(
                self.folder_path,
                f"universe_entity_{i}.txt"
            )

            with open(file_path, "w") as f:
                f.write(f"Entity content {i}")

        # Get the list of files in the folder
        files = [
            f for f in os.listdir(self.folder_path)
            if os.path.isfile(os.path.join(self.folder_path, f))
        ]

        initial_count = len(files)

        print(f"[Step 3] Initial file count: {initial_count}")

        # Calculate and delete half of the files randomly
        half_count = initial_count // 2
        files_to_delete = random.sample(files, half_count)

        for file_name in files_to_delete:
            os.remove(os.path.join(self.folder_path, file_name))

        # Count the remaining files
        remaining_files = [
            f for f in os.listdir(self.folder_path)
            if os.path.isfile(os.path.join(self.folder_path, f))
        ]

        final_count = len(remaining_files)

        print(
            f"[Step 5] Deleted {half_count} files. "
            f"Remaining file count: {final_count}"
        )

        return initial_count, final_count


# Create an object and run the task
if __name__ == "__main__":
    target_folder = "thanos_universe"

    thanos = ThanosSnap(
        target_folder,
        num_files_to_create=50
    )

    result = thanos.thanos_snap()

    print(f"Initial and final file counts: {result}")
```
