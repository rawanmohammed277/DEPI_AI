## **Problem 1: Two Sum**


```python
class Solution:

    def two_sum(self, numbers, target):
        number_indices = {}

        for index, current_number in enumerate(numbers):
            required_number = target - current_number

            if required_number in number_indices:
                return [number_indices[required_number], index]

            number_indices[current_number] = index
            
solution = Solution()

numbers = [2, 7, 11, 15]
target = 9

print(solution.two_sum(numbers, target))
```

## **Problem 2: Add Two Numbers** 

```python
class ListNode:
    def __init__(self, value=0, next_node=None):
        self.val = value
        self.next = next_node


class Solution:

    def addTwoNumbers(self, first_list, second_list):

        dummy_node = ListNode(0)
        current_node = dummy_node
        carry = 0

        while first_list or second_list or carry:

            first_value = first_list.val if first_list else 0
            second_value = second_list.val if second_list else 0

            total = first_value + second_value + carry

            digit = total % 10
            carry = total // 10

            current_node.next = ListNode(digit)
            current_node = current_node.next

            if first_list:
                first_list = first_list.next

            if second_list:
                second_list = second_list.next

        return dummy_node.next


first_list = ListNode(2)
first_list.next = ListNode(4)
first_list.next.next = ListNode(3)

second_list = ListNode(5)
second_list.next = ListNode(6)
second_list.next.next = ListNode(4)




solution = Solution()
result = solution.addTwoNumbers(first_list, second_list)

while result:
    print(result.val, end=" → ")
    result = result.next
```
