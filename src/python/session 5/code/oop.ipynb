```python
from abc import ABC , abstractmethod
class Animal(ABC):
    @abstractmethod
    def move(self):
        pass
    
class Bird(Animal):
    def move(self):
        print("move bird ")
    
class Cat(Animal):
    def move(self):
        print("move cat")
        
a=Bird()
a.move()

b=Cat()
b.move()    
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name 
        self.salary = salary
        
    def set_salary(self, obj, new_sal):
        obj.salary = new_sal  


class Manager:
    def __init__(self, name, salary):
        self.name = name 
        self.salary = salary

E1 = Employee("Depi", 2000)
m1 = Manager("Mohamed", 5000)

E1.set_salary(m1, 6000)
print(m1.salary)  
```
