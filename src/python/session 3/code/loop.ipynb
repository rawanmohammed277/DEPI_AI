## **-- break**
هطلع بره اللوب خالص 

## **-- continue**
هسكيب اللفه اللي فيها 

## **-- pass**
ملهاش تأثير

```python
for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
for i in range(10):
    if i==5:
        pass
    print(i)
```
