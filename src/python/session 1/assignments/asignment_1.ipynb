### **Task 1**

Emil: Amit_ml@gmail.edu

 Input Validation: Check if the input string contains exactly one "@" symbol and 
at least one "." after the "@" symbol. If it's not a valid email, return "Invalid 
email". 
 
- Extract Username: Extract and return the part of the email before the "@" 
symbol. 
 
- Extract Domain: Extract and return the domain (the part between "@" and the 
last "."). 
 
- Check for Domain Ending: Check if the email ends with ".com". If it does, 
return "Commercial Domain". If it ends with ".edu", return "Educational Domain". 
Otherwise, return "Other Domain". 

```python
email='Amit_ml@gmail.edu'
```

```python
if email.count("@") != 1 or "." not in email.split("@")[1]:
    print("Invalid email")
else:
    # Extract Username
    username = email[:email.index("@")]

    # Extract Domain
    domain = email[email.index("@") + 1 : email.rfind(".")]

    # Check Domain Ending
    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"

    print("Valid email")
    print("Username:", username)
    print("Domain:", domain)
    print(domain_type)
```

### **Task 3**

Encoded Message: &&&**$gnirtS PLIO!!@1234 
Steps to Decode: 
1. Extract the core part of the message: "gnirtS PLIO". 
2. Reverse the first word: "gnirtS" becomes "String". 
3. Replace shifted vowels in the second word: 
o "PLIO": Replace I->E and O->U to get "PLEU". 
4. Final decoded message: "String PLEU". 

```python
Encoded_Message2='&&&**$gnirtS PLIO!!@1234'
```

```python
# 1- Extract the core part
print("the core part: ",Encoded_Message2[6:17])

# 2-  Reverse the first word---------------------------------------
first_word2=(Encoded_Message2[6:12])
first_reserved_word2=first_word2[::-1]
print('Reverse of the first word: ',first_reserved_word2)

# 3- Replace sgifted vowels----------------------------------------
second_word2=(Encoded_Message2[13:17])
second_word2=second_word2.replace('I','E')
second_word2=second_word2.replace('O','U')
print("Replace shifted vowels: ", second_word2)

# 4- final decoded message---------------------------------------
print("Final decoded message: ",first_reserved_word2+' '+second_word2)
```

2. Reverse the first word: "yalpstcejorp" becomes "projectplay". 
3. Replace shifted vowels in the second word:

```python
# 1- Extract the core part
print("the core part: ",Encoded_Message3[6:24])

# 2-  Reverse the first word---------------------------------------
first_word3=(Encoded_Message3[6:18])
first_reserved_word3=first_word3[::-1]
first_reserved_word3_del=first_reserved_word3[:7]+first_reserved_word3[8:]
print('Reverse of the first word: ',first_reserved_word3_del)


# 3- Replace sgifted vowels----------------------------------------
second_word3=(Encoded_Message3[18:24])
second_word3=second_word3.replace('E','A')
second_word3=second_word3.replace('U','O')
print("Replace shifted vowels: ", second_word3)

# 4- final decoded message------------------------------------------
second_word3=second_word3.replace('OVT','TOV')

print("Final decoded message: ",first_reserved_word3_del+' '+second_word3)
```
