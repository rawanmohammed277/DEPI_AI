```python
def factorial(n: int):
    '''
    this function returns the factorial of a number n
    
    '''
    if n<0:
        return "factorial is not defined for negative numbers"
    else:
        if n==0 or n==1:
            return 1
        else:
            return n*factorial(n-1)
```

```python
def is_prime(n: int): # type hinting
    '''
    Check if a number is prime.
    
    Args:
        n (int): The number to check.
    
    Returns:
        bool: True if the number is prime, False otherwise.
    '''
    if n < 2:
        raise ValueError("Input must be an integer greater than or equal to 2.")
        
    if n > 1:
        for i in range(2, n):
            if n % i == 0:
                return False
        return True
    else:
        return False
```

```python
def common_divisor(a: int, b: int) -> int:
    '''
    Calculate the greatest common divisor (GCD) of two numbers using the Euclidean algorithm.
    
    Args:
        a (int): The first number.
        b (int): The second number.
    
    Returns:
        int: The greatest common divisor of a and b.
    '''
    
    limit = min(a, b)
    divisors=[]
    for divisor in range(1,limit+1):
        if a % divisor == 0 and b % divisor ==0:
            divisors.extend([divisor])
    return divisors


```

```python
def password_generator(length: int = 8) -> str:
    '''
    Generate a random password of a given length.
    
    Args:
        length (int): The length of the password to generate. Default is 8.
    
    Returns:
        str: A randomly generated password.
    '''
    char ='kclihbchbevciuepuiuevyedechgvwkGCVUGVECLUEVCLUJCHEGCUWE5#@%^&*'
    password = ''
    for _ in range(length):
        password += random.choice(char)
    return password
```

```python
import secrets
import string

def len_secure_password(length: int) -> str:
    '''
    Generate a secure password of a given length.
    
    Args:
        length (int): The length of the password to generate.
    
    Returns:
        str: A randomly generated secure password.
    '''
    if length <= 0:
        raise ValueError("Password length should be a positive integer.")
    
    char = (
        string.ascii_letters + string.digits + string.punctuation
        )
    return "".join(secrets.choice(char) for _ in range(length))
```

```python
len_secure_password(12)  # Generate a secure password of length 12
```

# **Chatbot**

```python
import random

# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}


# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
        
    return random.choice(responses["default"])

# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")

    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)

        if user_input == "goodbye":
            break

# Run the chatbot
chatbot()
```
