### **Q.1  Write a function that inputs a number and prints the multiplication table of that number**

```python
def multiplication_table():
    
    number=int(input("Enter a number: "))
    for i in range(1, 11):
        print(f"{number} x {i} = {number * i}")
        
multiplication_table()
```

## **Q2. Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes**


```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def print_twin_primes(limit=1000):
    print(f"Twin primes less than {limit}:")
    for p in range(3, limit - 2, 2):
        if is_prime(p) and is_prime(p + 2):
            print(f"({p}, {p + 2})")

print_twin_primes(1000)
```

## **Q.3 Write a program to find out the prime factors of a number. Example: prime factors of 56 - 2, 2, 2, 7**

```python
def get_prime_factors(n):
    factors = []
    while n % 2 == 0:
        factors.append(2)
        n //= 2
        
    divisor = 3
    while divisor * divisor <= n:
        while n % divisor == 0:
            factors.append(divisor)
            n //= divisor
        divisor += 2
        
    if n > 1:
        factors.append(n)
        
    return factors

try:
    number = int(input("Enter a number to find its prime factors: "))
    if number <= 1:
        print("Please enter a number greater than 1.")
    else:
        result = get_prime_factors(number)
        print(f"The prime factors of {number} are: {', '.join(map(str, result))}")
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

## **Q4. Write a function that converts a decimal number to binary number**

```python
def binary(number):
    if number == 0:
        return "0"
        
    reminders = []
    while number > 0:
        rem = number % 2
        reminders.append(str(rem))  
        number //= 2
        
    return ''.join(reminders[::-1])

num = int(input("Enter a number: "))
print(binary(num))
```

## **Q.5 A number is called perfect if the sum of proper divisors of that number is equal to the number. For example 28 is perfect number, since 1+2+4+7+14=28. Write a program to print all the perfect numbers in a given range**

```python
def is_perfect(n):
    if n < 2:
        return False
    
    divisors_sum = 1
    
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            divisors_sum += i
            if i != n // i:
                divisors_sum += n // i
                
    return divisors_sum == n

def find_perfect_numbers(start, end):
    perfect_numbers = []
    for num in range(start, end + 1):
        if is_perfect(num):
            perfect_numbers.append(num)
    return perfect_numbers

start_range = int(input("Enter start of range: "))
end_range = int(input("Enter end of range: "))

result = find_perfect_numbers(start_range, end_range)
print(f"Perfect numbers between {start_range} and {end_range}: {result}")
```
