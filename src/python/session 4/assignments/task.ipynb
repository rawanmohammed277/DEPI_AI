# **Basic Calculator**

```python
def get_number(prompt_message):
    """
    Prompts the user for numerical input and handles invalid non-numeric entries.

    :param prompt_message: The message displayed to the user
    :type prompt_message: str

    :return: Validated floating-point number
    :rtype: float
    """
    while True:
        try:
            return float(input(prompt_message))
        except ValueError:
            print("Invalid input! Please enter a valid number.")


def calculate(op_choice, num1, num2):
    """
    Performs the specified mathematical operation on two numbers.

    :param op_choice: Menu selection string ('1', '2', '3', or '4')
    :type op_choice: str
    :param num1: The first number
    :type num1: float
    :param num2: The second number
    :type num2: float

    :raise ZeroDivisionError: If division by zero is attempted

    :return: A tuple containing the calculated result and operation label
    :rtype: tuple
    """
    if op_choice == '1':
        return num1 + num2, "adding"
    elif op_choice == '2':
        return num1 - num2, "subtracting"
    elif op_choice == '3':
        return num1 * num2, "multiplying"
    elif op_choice == '4':
        if num2 == 0:
            raise ZeroDivisionError("Cannot divide by zero.")
        return num1 / num2, "dividing"


def format_num(val):
    """
    Formats floating-point numbers to integers if they have no fractional part.

    :param val: Numerical value to format
    :type val: float

    :return: Formatted integer or float
    :rtype: int or float
    """
    return int(val) if val.is_integer() else val


def main():
    """
    Runs the main calculator loop, managing menu navigation and user interaction.

    :return: None
    :rtype: None
    """
    while True:
        print("\n=== Basic Calculator ===")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Exit")
        
        choice = input("Select an option (1-5): ").strip()
        
        if choice == '5':
            print("Thank you for using the calculator! Goodbye.")
            break
            
        if choice not in ['1', '2', '3', '4']:
            print("Invalid selection! Please choose an option between 1 and 5.")
            continue
            
        n1 = get_number("Enter the first number: ")
        n2 = get_number("Enter the second number: ")
        
        try:
            result, operation_name = calculate(choice, n1, n2)
            print(f"The result of {operation_name} {format_num(n1)} and {format_num(n2)} is {format_num(result)}.")
        except ZeroDivisionError:
            print("Error: Division by zero is not allowed.")
            
        again = input("\nDo you want to perform another calculation? (yes/no): ").strip().lower()
        if again not in ['yes', 'y']:
            print("Thank you for using the calculator! Goodbye.")
            break


if __name__ == "__main__":
    main()
```

```python
import os
import random


def thanos_snap(folder_path, num_files_to_create=50):
    """
    Simulates the 'Thanos Snap' by creating files in a specified directory,
    checking total file count, and randomly deleting exactly half of them.

    :param folder_path: The directory path to perform the operation in
    :type folder_path: str
    :param num_files_to_create: Number of dummy files to create
    :type num_files_to_create: int

    :raise OSError: If directory creation or file deletion fails

    :return: A tuple containing (initial_file_count, remaining_file_count)
    :rtype: tuple
    """
    # 1. Open/Ensure the target folder exists
    os.makedirs(folder_path, exist_ok=True)

    # 2. Create the specified number of files inside it
    for i in range(1, num_files_to_create + 1):
        file_path = os.path.join(folder_path, f"universe_entity_{i}.txt")
        with open(file_path, "w") as f:
            f.write(f"Entity content {i}")

    # 3. Check the length of files in that folder
    files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    initial_count = len(files)
    print(f"[Step 3] Initial file count: {initial_count}")

    # 4. Delete random half of this folder
    half_count = initial_count // 2
    files_to_delete = random.sample(files, half_count)

    for file_name in files_to_delete:
        os.remove(os.path.join(folder_path, file_name))

    # 5. Check the result
    remaining_files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    final_count = len(remaining_files)
    print(f"[Step 5] Deleted {half_count} files. Remaining file count: {final_count}")

    return initial_count, final_count


# Execution example
if __name__ == "__main__":
    target_folder = "thanos_universe"
    thanos_snap(target_folder, num_files_to_create=50)
```
