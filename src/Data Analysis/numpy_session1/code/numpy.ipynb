```python
def array_reshape(arr, row= -1, cols= -1):
    try:
        arr= np.array(arr)
        return arr.reshape((row, cols))
    except Exception as e:
        raise ValueError("Invalid Reshape Operation:{e}")
```

```python
vec= [1,2,3,4,5,6]
print(array_reshape(vec,2,3))
```
