```python
# plt.plot(name,age)
# plt.title('Name and Age')
# plt.xlabel('Name')
# plt.ylabel('Age')
# plt.savefig('students.png')
x= np.linspace(0,10,100) #   نبدا من 0 لحد 100  ستارت و ايند 
y= np.sin(x)
#plt.plot(x,y,'o-')
plt.plot(x,y,linestyle='-.',color='r',linewidth=2)
plt.title('Sin Wave')
plt.xlabel('X-axis')
plt.ylabel('sin(x)')
plt.grid(True,color='g',alpha=0.8)
plt.xlim(2,8)
plt.ylim(-0.5,1)
plt.show()
```

```python
x= np.linspace(0,10,50)
y= np.sin(x) + np.random.rand(50)
size= np.random.randint(50,500,50)
color= np.linspace(0,1,50)

plt.scatter(x,y,c=color,s=size,marker='*',alpha=0.8,cmap='viridis', edgecolors='black',linewidths=1.5)
plt.title('Scatter Plot',fontsize=12,c='brown')
plt.xlabel('X-Values')
plt.ylabel('Y-Values')
plt.colorbar(label='Color Scale')
plt.tight_layout()

plt.show()
```

```python
classes= ['A','B','C']
students= [30,45,20]

plt.bar(classes,students,width=0.5,color='skyblue',edgecolor='black',alpha=0.8)
for i,value in enumerate(students):
    plt.text(i, value+1, str(value), ha='center')

plt.title('Number of students in each class')
plt.xlabel('classes')
plt.ylabel('Number of students')


plt.show()
```

```python
fruites= ['Apple','Banana','Kiwi','Orange']
explode= [0.1,0,0.1,0]
sales= [30,15,50,20,]

plt.pie(sales, labels= fruites, autopct= "%1.3f%%",wedgeprops= {"width":0.4} )
plt.title('Fruit Sales')
plt.show()
```

```python
fruites= ['Apple','Banana','Kiwi','Orange']
explode= [0.1,0,0.1,0]
sales= [30,15,50,20,]

plt.pie(sales, labels= fruites, autopct= "%1.3f%%", explode=explode )
plt.title('Fruit Sales')
plt.show()
```

```python
np.random.seed(2)
scores= np.random.normal(loc= 75, scale= 10, size= 100)
scores= np.clip(scores, 0, 100)

plt.figure(figsize= (10,6))
plt.hist(scores, 5, color='skyblue', edgecolor='black')
plt.title('Distribution of students scores', fontsize=16)
plt.xlabel('Scores')
plt.ylabel('Students')
plt.grid(axis= 'y', linestyle='--',alpha=0.5)
plt.show()
```
