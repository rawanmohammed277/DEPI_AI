# Assignment 1:
### - The automated_stat_analyzer Function
- Scenario: A retail company needs a utility to quickly summarize sales data. Students must create a function that identifies the 
"Central Tendency" and "Dispersion" of any numerical column.
- ### Requirements:

* Accept a Pandas DataFrame and a column name.

* Calculate the Mean, Median, and Standard Deviation .

* Identify if the data is "Skewed" by comparing the Mean and Median.


* Bonus: If the column is categorical, return the Mode instead.

### Your Data

```python
import pandas as pd
import numpy as np

# Create a synthetic Company Sales Dataset
data = {
    'Transaction_ID': range(1, 11),
    'Product_Category': ['Electronics', 'Home', 'Electronics', 'Sports', 'Home', 
                         'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'],
    'Sales_Amount': [150, 200, 155, 300, 210, 180, 205, 1000, 190, 160], # 1000 is an Outlier
    'Customer_Age': [25, 34, np.nan, 45, 23, 31, 29, np.nan, 38, 40],    # Contains Nulls (NaN)
    'Rating': [5, 4, 3, 5, 2, 4, 5, 2, 4, 3]
}

df_test = pd.DataFrame(data)

# Save to CSV for students to practice loading files [cite: 74]
df_test.to_csv('company_sales_test.csv', index=False)
print("Test dataset created successfully!")
```

```python
df_test.head()
```

```python
def automated_stat_analyzer(df, column_name):
    # Check if column exists
    if column_name not in df.columns:
        raise ValueError(f"Column '{column_name}' does not exist.")

    column = df[column_name]
    
    # Numerical Column

    if pd.api.types.is_numeric_dtype(column):

        mean = column.mean()
        median = column.median()
        std = column.std()

        # Compare Mean and Median
        if mean > median:
            skewed = "Right Skewed"
        elif mean < median:
            skewed = "Left Skewed"
        else:
            skewed = "Not Skewed"

        return {
            "Column": column_name,
            "Type": "Numerical",
            "Mean": mean,
            "Median": median,
            "Standard Deviation": std,
            "Skewness": skewed
        }

    # Categorical Column

    else:

        mode = column.mode()

        return {
            "Column": column_name,
            "Type": "Categorical",
            "Mode": mode.tolist()
        }
```

```python
# TEST CASE 1: Sales Amount
print("\n===== Test 1: Sales_Amount =====")

result = automated_stat_analyzer(
    df_test,
    "Sales_Amount"
)

print(result)

# TEST CASE 2: Product Category

print("\n===== Test 2: Product_Category =====")

result = automated_stat_analyzer(
    df_test,
    "Product_Category"
)

print(result)

# TEST CASE 3: Customer Age

print("\n===== Test 3: Customer_Age =====")

result = automated_stat_analyzer(
    df_test,
    "Customer_Age"
)

print(result)


# TEST CASE 4: Rating

print("\n===== Test 4: Rating =====")

result = automated_stat_analyzer(
    df_test,
    "Rating"
)

print(result)
```

## Assignment 2: 
  ### The null_handling_strategy Function


#### Scenario: Incoming user data often has missing values.Students must implement a flexible strategy to handle these "Null Values" to prepare data for Machine Learning.
### Requirements:

* Check for null values in the DataFrame.

* Apply a strategy based on parameters: "drop_rows", "fill_mean", or "fill_median" .

* Ensure the function only fills numerical columns when using mean or median.

```python
import pandas as pd
import numpy as np


data = {
    'Transaction_ID': range(1, 11),

    'Product_Category': [
        'Electronics', 'Home', 'Electronics', 'Sports', 'Home',
        'Electronics', 'Home', 'Sports', 'Electronics', 'Electronics'
    ],

    'Sales_Amount': [
        150, 200, 155, 300, 210,
        180, 205, 1000, 190, 160
    ],

    'Customer_Age': [
        25, 34, np.nan, 45, 23,
        31, 29, np.nan, 38, 40
    ],

    'Rating': [
        5, 4, 3, 5, 2,
        4, 5, 2, 4, 3
    ]
}

df_test = pd.DataFrame(data)

# Null Handling Strategy


def null_handling_strategy(df, strategy="fill_mean"):
    """
    Company Task: Clean a dataset by resolving missing (NaN) values.

    Strategies:
    - drop_rows
    - fill_mean
    - fill_median
    """

    # Check if there are null values
    if not df.isnull().values.any():
        print("No null values found.")
        return df.copy()

    # Make a copy so the original DataFrame is not modified
    result = df.copy()

    # Strategy 1: Drop rows containing null values

    if strategy == "drop_rows":

        result = result.dropna()

    # Strategy 2: Fill numerical columns with Mean

    elif strategy == "fill_mean":

        numerical_columns = result.select_dtypes(
            include=np.number
        ).columns

        for column in numerical_columns:
            result[column] = result[column].fillna(
                result[column].mean()
            )

    # Strategy 3: Fill numerical columns with Median

    elif strategy == "fill_median":

        numerical_columns = result.select_dtypes(
            include=np.number
        ).columns

        for column in numerical_columns:
            result[column] = result[column].fillna(
                result[column].median()
            )

    # Invalid Strategy

    else:
        raise ValueError(
            "Invalid strategy. Choose: "
            "'drop_rows', 'fill_mean', or 'fill_median'."
        )

    return result

# TEST 1: fill_mean


print("===== Original Data =====")
print(df_test)

print("\n===== Test 1: Fill Mean =====")

result_mean = null_handling_strategy(
    df_test,
    "fill_mean"
)

print(result_mean)


# TEST 2: fill_median

print("\n===== Test 2: Fill Median =====")

result_median = null_handling_strategy(
    df_test,
    "fill_median"
)

print(result_median)


# TEST 3: drop_rows

print("\n===== Test 3: Drop Rows =====")

result_drop = null_handling_strategy(
    df_test,
    "drop_rows"
)

print(result_drop)


# TEST 4: Invalid Strategy

print("\n===== Test 4: Invalid Strategy =====")

try:
    result = null_handling_strategy(
        df_test,
        "fill_mode"
    )

    print(result)

except ValueError as e:
    print(e)
```
