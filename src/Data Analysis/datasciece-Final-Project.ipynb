

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with median for numeric and mode for categorical

```python
# trying to fill missing values in my datset using for loop 

for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col].fillna(df[col].median(), inplace=True)
    else:
        df[col].fillna('Unknown', inplace=True)
```

```python
#check the Nulls 
df.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 
df['member_gender'].value_counts()
print(df.member_gender.head(15))
```

```python
# fill  dataset with missing values.

df['member_gender'] = df['member_gender'].replace({'Unknown': 'Other'})    
```


# <a id="13">  13 ) . What is the  Correlation between each variable in the dataset?</a>

```python
# Select numerical features for correlation analysis
numeric_vars = df.select_dtypes(include=['float64', 'int64']).columns

# Plotting the correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(df[numeric_vars].corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numerical Variables', fontsize=14)
plt.show()
```



> ### <mark> 
- Perfect Correlation ($r = 1.00$):duration_sec, duration_min, and duration_hours show a correlation of $1.00$ because they represent the exact same measurement across different units.age and member_birth_year display a perfect inverse correlation ($r = -1.00$), as age is computed directly as $2019 - \text{birth\_year}$.
- Geographic/Spatial Correlations ($r \approx 0.99$):Strong positive correlation exists between start_station_latitude and end_station_latitude ($r = 0.99$), as well as between start_station_longitude and end_station_longitude ($r = 0.99$). This indicates that most bike trips stay within close geographical proximity.
- No Linear Relationship with Duration ($r \approx 0.00$):User age has virtually no linear correlation with trip duration ($r \approx 0.00$).Station IDs and Bike IDs show no meaningful linear relationship with trip length or user demographics.
 </mark>
